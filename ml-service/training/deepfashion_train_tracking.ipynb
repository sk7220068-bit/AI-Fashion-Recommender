### 📊 Experiment Tracking (Local/Managed)
Check your MLflow runs to see metrics, weights, and predictions.

```python
import mlflow
import os
from pathlib import Path

# Initialize MLflow
mlflow.set_experiment("DeepFashion2_FineTuning")
```

### 📦 Dataset Acquisition (Kaggle Native)
Instead of brittle Google Drive downloads, we now use Kaggle datasets directly.

**Instructions:**
1. Click **+ Add Data** in the right sidebar.
2. Search for `pavansanagapati/deepfashion2` or `shon721/deepfashion2`.
3. Click **Add**.
4. The path will usually be `/kaggle/input/deepfashion2/`.

```python
import glob

# Auto-detect the dataset path
kaggle_inputs = glob.glob("/kaggle/input/**/*", recursive=False)
print("Detected Inputs:", kaggle_inputs)

data_path = None
for p in kaggle_inputs:
    if "deepfashion2" in p.lower():
        data_path = p
        break

if data_path:
    print(f"✓ DeepFashion2 found at: {data_path}")
else:
    print("❌ DeepFashion2 dataset NOT found. Please add it via '+ Add Data' sidebar.")
```

### 🎯 Model Training
Starting from your custom weights to achieve maximum accuracy.

```python
from ultralytics import YOLO

# 1. Load your Kaggle weights as the base
# IF RUNNING ON KAGGLE: Either upload the file or mount it from a dataset
# Example: /kaggle/input/my-models/yolov8n.pt
model_path = "yolov8n.pt" # Path to your uploaded yolov8n.pt file
model = YOLO(model_path)

# 2. Train with MLflow tracking enabled
with mlflow.start_run(run_name="Custom_Fashion_FineTune"):
    results = model.train(
        data=f"{data_path}/data.yaml" if data_path else "./default_data.yaml",
        epochs=50,
        imgsz=640,
        batch=16,
        mosaic=1.0,
        mixup=0.2,
        project="fashion_project",
        name="fine_tune_results"
    )

print("Training Complete. Optimized model saved in 'fashion_project/fine_tune_results/weights/best.pt'")
```
